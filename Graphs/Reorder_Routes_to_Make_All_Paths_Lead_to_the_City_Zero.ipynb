```python
from typing import List

class Solution:

    # DFS function to traverse graph
    def dfs(self, node, forward, backward, vis):
        vis.add(node)  # mark current node as visited
        changes = 0    # count of edges to reverse

        # Traverse original directed edges (u -> v)
        for nei in forward[node]:
            if nei not in vis:
                changes += 1  # this edge is in wrong direction for root (0)
                changes += self.dfs(nei, forward, backward, vis)

        # Traverse reverse edges (v -> u), no reversal needed
        for nei in backward[node]:
            if nei not in vis:
                changes += self.dfs(nei, forward, backward, vis)

        return changes

    def minReorder(self, n: int, connections: List[List[int]]) -> int:
        vis = set()

        # adjacency list for original direction
        forward = [[] for _ in range(n)]

        # adjacency list for reverse direction
        backward = [[] for _ in range(n)]

        # build graph
        for u, v in connections:
            forward[u].append(v)
            backward[v].append(u)

        # start DFS from node 0
        return self.dfs(0, forward, backward, vis)
```

```python
# Example input
n = 6
connections = [[0,1],[1,3],[2,3],[4,0],[4,5]]

sol = Solution()   # DFS Solution class
result = sol.minReorder(n, connections)

print("DFS Answer:", result)
```

```python
from typing import List
from collections import deque

class Solution:

    def minReorder(self, n: int, connections: List[List[int]]) -> int:
        # adjacency lists
        forward = [[] for _ in range(n)]
        backward = [[] for _ in range(n)]

        # build graph
        for u, v in connections:
            forward[u].append(v)
            backward[v].append(u)

        vis = set()
        q = deque([0])   # start BFS from node 0
        vis.add(0)

        changes = 0

        while q:
            node = q.popleft()

            # forward edges (u -> v) need reversal
            for nei in forward[node]:
                if nei not in vis:
                    changes += 1
                    vis.add(nei)
                    q.append(nei)

            # backward edges (v -> u) are fine
            for nei in backward[node]:
                if nei not in vis:
                    vis.add(nei)
                    q.append(nei)

        return changes
```

```python
from typing import List

# Example input
n = 6
connections = [[0,1],[1,3],[2,3],[4,0],[4,5]]

sol = Solution()   # BFS Solution class
result = sol.minReorder(n, connections)

print("BFS Answer:", result)
```
