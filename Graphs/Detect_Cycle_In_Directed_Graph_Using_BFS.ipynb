```python
from collections import deque


# TC: O(V + E)
# SC: O(V + E)

# APPROACH:
# 1. Calculate indegree of every node.
# 2. Put all nodes having indegree 0 into queue.
# 3. Pop node from queue using BFS.
# 4. Add node into topological order.
# 5. Reduce indegree of neighbor nodes.
# 6. If neighbor indegree becomes 0, push into queue.
# 7. If topo size != totalNodes -> cycle exists.


class Solution:

    # Kahn's Algorithm (BFS Topological Sort)
    def kahnsAlgo(self, totalNodes, graph):

        # store indegree of every node
        indegree = [0] * totalNodes

        # calculate indegree
        for currentNode in range(totalNodes):

            for neighborNode in graph[currentNode]:

                indegree[neighborNode] += 1

        # queue for BFS
        queue = deque()

        # add all indegree 0 nodes
        for node in range(totalNodes):

            if indegree[node] == 0:
                queue.append(node)

        # store topological order
        topo = []

        # BFS
        while queue:

            # remove front node
            currentNode = queue.popleft()

            # add into answer
            topo.append(currentNode)

            # visit neighbors
            for neighborNode in graph[currentNode]:

                # reduce indegree
                indegree[neighborNode] -= 1

                # if indegree becomes 0
                if indegree[neighborNode] == 0:
                    queue.append(neighborNode)

        # cycle detection
        if len(topo) != totalNodes:
            print("Cycle Exists")
        else:
            print("No Cycle")

        return topo


# Driver Code
totalNodes = 6

graph = [
    [2, 3],   # 0
    [3, 4],   # 1
    [5],      # 2
    [5],      # 3
    [],       # 4
    []        # 5
]

obj = Solution()

print(obj.kahnsAlgo(totalNodes, graph))
```
