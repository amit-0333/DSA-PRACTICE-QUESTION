```python
# TC: O(V + E)
# SC: O(V)

# APPROACH:
# 1. Use DFS traversal.
# 2. Visit all neighbors first.
# 3. Push node into stack after DFS call.
# 4. Pop stack to get topological order.


def topSortDFS(currentNode, graph, visited, stack):

    # mark visited
    visited.add(currentNode)

    # visit neighbors
    for neighborNode in graph[currentNode]:

        if neighborNode not in visited:

            topSortDFS(neighborNode, graph, visited, stack)

    # push into stack
    stack.append(currentNode)


def topologicalSort(totalNodes, graph):

    visited = set()

    stack = []

    # visit all nodes
    for node in range(totalNodes):

        if node not in visited:

            topSortDFS(node, graph, visited, stack)

    # print topo order
    while stack:

        node = stack.pop()

        print(node, end=" ")


# Driver Code
totalNodes = 6

graph = [
    [2, 3],   # 0
    [3, 4],   # 1
    [5],      # 2
    [5],      # 3
    [],       # 4
    []        # 5
]

topologicalSort(totalNodes, graph)
```
