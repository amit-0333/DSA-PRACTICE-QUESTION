```python
from collections import deque

# APPROACH:
# 1. Push all rotten oranges into queue initially.
# 2. Perform multi-source BFS.
# 3. Process oranges level by level.
# 4. If fresh neighbour exists:
#       - rot it
#       - push into queue
# 5. Increase time after each level.
# 6. If fresh orange remains, return -1.

# TC: O(n * m)
# SC: O(n * m)

class Solution:

    def orangesRotting(self, grid):

        # Directions: down, up, right, left
        direction = [(1,0), (-1,0), (0,1), (0,-1)]

        n = len(grid)
        m = len(grid[0])

        # Queue for multi-source BFS
        q = deque()

        # Push all rotten oranges
        for i in range(n):
            for j in range(m):

                if grid[i][j] == 2:
                    q.append((i, j))

        time = 0

        # BFS traversal
        while q:

            size = len(q)

            # To check whether new oranges rot at this level
            temp = 0

            while size > 0:

                r, c = q.popleft()

                # Visit neighbours
                for dr, dc in direction:

                    nr = r + dr
                    nc = c + dc

                    # Valid fresh orange
                    if 0 <= nr < n and 0 <= nc < m and grid[nr][nc] == 1:

                        # Rot orange
                        grid[nr][nc] = 2

                        temp = 1

                        q.append((nr, nc))

                size -= 1

            # Increase time only if some orange rotted
            time += temp

        # Check for remaining fresh oranges
        for i in range(n):
            for j in range(m):

                if grid[i][j] == 1:
                    return -1

        return time


# ================= DRIVER CODE =================

grid = [
    [2,1,1],
    [1,1,0],
    [0,1,1]
]

obj = Solution()

result = obj.orangesRotting(grid)

print("Minimum time:", result)
```
