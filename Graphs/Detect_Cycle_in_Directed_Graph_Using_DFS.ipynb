```python
# TC: O(V + E)
# SC: O(V)

# APPROACH:
# 1. visited -> globally visited nodes.
# 2. path -> current recursion path.
# 3. If node appears again in current path,
#    cycle exists.


def dfs(node, graph, visited, path):

    # mark visited
    visited.add(node)

    # add into current path
    path.add(node)

    # visit neighbours
    for nei in graph[node]:

        # if not visited
        if nei not in visited:

            if dfs(nei, graph, visited, path):
                return True

        # node already in current path -> cycle
        elif nei in path:
            return True

    # backtracking
    path.remove(node)

    return False

graph = {
    0: [1],
    1: [2],
    2: [0]
}

visited = set()
path = set()

print(dfs(0, graph, visited, path))
```
