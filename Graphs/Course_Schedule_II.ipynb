```python
from collections import deque
from typing import List


# TC: O(V + E)
# SC: O(V + E)

# APPROACH:
# 1. Create graph from prerequisites.
# 2. Calculate indegree of every course.
# 3. Push all indegree 0 courses into queue.
# 4. Perform BFS (Kahn's Algorithm).
# 5. Add course into answer.
# 6. Reduce indegree of neighbors.
# 7. If indegree becomes 0, push into queue.
# 8. If all courses processed -> return topo order.
# 9. Else return [] because cycle exists.


class Solution:

    def findOrder(self, numCourses: int, prerequisites: List[List[int]]) -> List[int]:

        # graph
        graph = [[] for _ in range(numCourses)]

        # indegree array
        indegree = [0] * numCourses

        # build graph
        for course, prerequisite in prerequisites:

            graph[prerequisite].append(course)

            indegree[course] += 1

        # BFS queue
        queue = deque()

        # push indegree 0 nodes
        for course in range(numCourses):

            if indegree[course] == 0:

                queue.append(course)

        # store topo order
        topo = []

        # BFS
        while queue:

            currentCourse = queue.popleft()

            topo.append(currentCourse)

            # visit neighbors
            for neighborCourse in graph[currentCourse]:

                indegree[neighborCourse] -= 1

                # if indegree becomes 0
                if indegree[neighborCourse] == 0:

                    queue.append(neighborCourse)

        # cycle check
        if len(topo) == numCourses:

            return topo

        return []
```
