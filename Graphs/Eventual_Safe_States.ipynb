```python
# TC : O(V + E)
# SC : O(V)

# APPROACH
# path[] is being used as:
# True  -> node is unsafe / part of current recursion cycle
# False -> safe node
#
# This works for many cases because nodes inside cycles
# return early before backtracking resets them to False.
#
# BUT this logic fails for nodes that are not directly
# in cycle but can reach a cycle.


def dfs(node, graph, vis, path):

    vis.add(node)
    path[node] = True

    for nei in graph[node]:

        # If neighbor not visited → DFS
        if nei not in vis:

            if dfs(nei, graph, vis, path):
                return True

        # If neighbor already in current path → cycle
        elif path[nei]:
            return True

    # Backtracking
    path[node] = False

    return False


def eventualSeqSteps(V, edges):

    path = [False] * V
    vis = set()
    res = []

    for i in range(V):

        if i not in vis:
            dfs(i, edges, vis, path)

    for i in range(V):

        if path[i] == False:
            res.append(i)

    return res


# ---------------- DRIVER ----------------

graph = [
    [1],   # 0
    [2],   # 1
    [3],   # 2
    [1],   # 3 -> cycle 1-2-3
    [0]    # 4 reaches cycle
]

V = 5

print(eventualSeqSteps(V, graph))
```

```python
# TC : O(V + E)
# SC : O(V)

# APPROACH
# vis[]    -> globally visited nodes
# path[]   -> current recursion stack
# unsafe[] -> stores nodes that are part of / reach cycle
#
# If a node:
# - is inside a cycle
# - OR can reach a cycle
#
# then unsafe[node] = True


def dfs(node, graph, vis, path, unsafe):

    vis.add(node)
    path[node] = True

    for nei in graph[node]:

        # DFS on unvisited node
        if nei not in vis:

            if dfs(nei, graph, vis, path, unsafe):
                unsafe[node] = True
                return True

        # Cycle detected
        elif path[nei]:
            unsafe[node] = True
            return True

        # Neighbor already unsafe
        elif unsafe[nei]:
            unsafe[node] = True
            return True

    # Backtracking
    path[node] = False

    return False


def eventualSeqSteps(V, edges):

    path = [False] * V
    unsafe = [False] * V

    vis = set()
    res = []

    for i in range(V):

        if i not in vis:
            dfs(i, edges, vis, path, unsafe)

    for i in range(V):

        # Safe nodes
        if unsafe[i] == False:
            res.append(i)

    return res


# ---------------- DRIVER ----------------

graph = [
    [1],   # 0
    [2],   # 1
    [3],   # 2
    [1],   # 3 -> cycle 1-2-3
    [0]    # 4 reaches cycle
]

V = 5

print(eventualSeqSteps(V, graph))
```
