```python
import heapq


# TC: O((V + E) log V)
# SC: O(V)

# APPROACH:
# 1. Store shortest distance of every node.
# 2. Use min heap to get minimum distance node.
# 3. Relax neighbor distances.
# 4. Update heap when shorter path found.


def dijkstra(graph, startVertex):

    # distance dictionary
    distances = {
        vertex: float('inf') for vertex in graph
    }

    distances[startVertex] = 0

    # min heap -> (distance, node)
    minHeap = [(0, startVertex)]

    # Dijkstra
    while minHeap:

        currentDistance, currentVertex = heapq.heappop(minHeap)

        # skip old larger distance
        if currentDistance > distances[currentVertex]:
            continue

        # visit neighbors
        for neighborNode, weight in graph[currentVertex]:

            newDistance = currentDistance + weight

            # relaxation
            if newDistance < distances[neighborNode]:

                distances[neighborNode] = newDistance

                heapq.heappush(
                    minHeap,
                    (newDistance, neighborNode)
                )

    return distances


# Driver Code
graph = {
    0: [(1, 2), (2, 4)],
    1: [(0, 2), (2, 1), (3, 7)],
    2: [(0, 4), (1, 1), (4, 3)],
    3: [(1, 7), (4, 1)],
    4: [(2, 3), (3, 1)]
}

print(dijkstra(graph, 0))
```
