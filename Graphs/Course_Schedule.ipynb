```python
from typing import List


# TC: O(V + E)
# SC: O(V + E)

# APPROACH:
# 1. Build directed graph.
# 2. Use DFS traversal.
# 3. Maintain recursion stack (path).
# 4. If a node appears again in path -> cycle exists.


class Solution:

    def dfs(self, node, graph, visited, path):

        visited.add(node)
        path.add(node)

        for nei in graph[node]:

            # not visited
            if nei not in visited:

                if self.dfs(nei, graph, visited, path):
                    return True

            # already in current path → cycle
            elif nei in path:
                return True

        # backtrack
        path.remove(node)

        return False

    def canFinish(self, numCourses: int, prerequisites: List[List[int]]) -> bool:

        # build graph
        graph = {i: [] for i in range(numCourses)}

        for course, prereq in prerequisites:
            graph[prereq].append(course)

        visited = set()
        path = set()

        # check all components
        for node in range(numCourses):

            if node not in visited:

                if self.dfs(node, graph, visited, path):
                    return False

        return True

# Example usage

sol = Solution()

numCourses = 4
prerequisites = [[1, 0], [2, 1], [3, 2]]

print(sol.canFinish(numCourses, prerequisites))
```
