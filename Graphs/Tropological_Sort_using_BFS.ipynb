```python
from collections import deque


# TC: O(V + E)
# SC: O(V)

# APPROACH:
# 1. Calculate indegree of every node.
# 2. Push all indegree 0 nodes into queue.
# 3. Perform BFS.
# 4. Remove node from queue and add to topo order.
# 5. Reduce indegree of neighbors.
# 6. If indegree becomes 0, push into queue.


def topologicalSort(totalNodes, graph):

    # indegree array
    indegree = [0] * totalNodes

    # calculate indegree
    for currentNode in range(totalNodes):

        for neighborNode in graph[currentNode]:

            indegree[neighborNode] += 1

    # BFS queue
    queue = deque()

    # push indegree 0 nodes
    for node in range(totalNodes):

        if indegree[node] == 0:
            queue.append(node)

    # store answer
    topo = []

    # BFS
    while queue:

        currentNode = queue.popleft()

        topo.append(currentNode)

        # visit neighbors
        for neighborNode in graph[currentNode]:

            indegree[neighborNode] -= 1

            # if indegree becomes 0
            if indegree[neighborNode] == 0:

                queue.append(neighborNode)

    print(topo)


# Driver Code
totalNodes = 6

graph = [
    [2, 3],   # 0
    [3, 4],   # 1
    [5],      # 2
    [5],      # 3
    [],       # 4
    []        # 5
]

topologicalSort(totalNodes, graph)
```
