```python
# TC : O(N)
# SC : O(N)

# APPROACH
# vis  -> globally visited nodes
# path -> stores DFS depth of nodes in current recursion stack
#
# If we revisit a node already in current path,
# cycle length =
# current_depth - previous_depth + 1


class Solution:

    def __init__(self):

        # Store maximum cycle length
        self.longestCycleLen = -1

    def dfs(self, node, graph, vis, path, cyclelen):

        # Mark node visited globally
        vis.add(node)

        # Store current DFS depth
        path[node] = cyclelen

        # Current node has only one outgoing edge
        nei = graph[node]

        # Continue only if edge exists
        if nei != -1:

            # If neighbor not visited → DFS
            if nei not in vis:

                self.dfs(
                    nei,
                    graph,
                    vis,
                    path,
                    cyclelen + 1
                )

            # If neighbor already in current path
            # then cycle detected
            elif path[nei] != -1:

                # Calculate cycle length
                currlen = cyclelen - path[nei] + 1

                # Update maximum cycle length
                self.longestCycleLen = max(
                    self.longestCycleLen,
                    currlen
                )

        # Backtracking
        # Remove node from current DFS path
        path[node] = -1

    def longestCycle(self, edges):

        n = len(edges)

        # Globally visited nodes
        vis = set()

        # -1 means node not in current path
        path = [-1] * n

        # Try DFS from every node
        for i in range(n):

            if i not in vis:

                self.dfs(
                    i,
                    edges,
                    vis,
                    path,
                    0
                )

        return self.longestCycleLen


# ---------------- DRIVER CODE ----------------

test_cases = [

    [3, 3, 4, 2, 3],      # Answer = 3
    [2, -1, 3, 1],        # Answer = -1
    [1, 2, 0],            # Answer = 3
    [1, 2, 3, 4, 0],      # Answer = 5
    [-1, -1, -1],         # Answer = -1

]


sol = Solution()

for idx, edges in enumerate(test_cases, 1):

    # Reset answer for every testcase
    sol.longestCycleLen = -1

    print(f"Test Case {idx}")

    print("Edges :", edges)

    ans = sol.longestCycle(edges)

    print("Longest Cycle Length :", ans)

    print("-" * 40)
```
