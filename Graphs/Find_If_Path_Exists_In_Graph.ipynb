```python
from typing import List

# TC: O(V + E)
# SC: O(V + E)

# APPROACH:
# 1. Create adjacency list from edges.
# 2. Start DFS from source node.
# 3. Mark all reachable nodes as visited.
# 4. If destination is visited -> path exists.

class Solution:

    # DFS traversal
    def dfs(self, node, graph, vis):

        # mark node visited
        vis.add(node)

        # visit all neighbors
        for nei in graph[node]:

            # if neighbor not visited
            if nei not in vis:

                # recursive DFS call
                self.dfs(nei, graph, vis)

    def validPath(self, n: int, edges: List[List[int]],
                  source: int, destination: int) -> bool:

        # visited set
        vis = set()

        # adjacency list
        graph = [[] for _ in range(n)]

        # build graph
        for u, v in edges:
            graph[u].append(v)
            graph[v].append(u)

        # start DFS from source
        self.dfs(source, graph, vis)

        # check destination reachable
        return destination in vis

# ---------------- DRIVER CODE ----------------

n = 6

edges = [
    [0, 1],
    [0, 2],
    [3, 5],
    [5, 4],
    [4, 3]
]

source = 0
destination = 5

obj = Solution()

ans = obj.validPath(n, edges, source, destination)

print(ans)
```
