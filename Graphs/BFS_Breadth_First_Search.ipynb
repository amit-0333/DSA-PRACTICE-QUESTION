```python
from collections import deque

# TC: O(V + E)
# SC: O(V)

# APPROACH:
# 1. Use queue for level order traversal.
# 2. Mark starting node visited.
# 3. Pop node from queue.
# 4. Push all unvisited neighbours into queue.


def bfs(node, graph, visited):

    # create queue
    q = deque()

    # mark starting node visited
    visited.add(node)

    # push into queue
    q.append(node)

    while q:

        # remove front node
        node = q.popleft()

        # process node
        print(node, end=" ")

        # visit neighbours
        for adj_node in graph[node]:

            # visit only unvisited nodes
            if adj_node not in visited:

                visited.add(adj_node)

                q.append(adj_node)


# ---------------- DRIVER CODE ----------------

graph = {
    1: [2, 3],
    2: [4, 5],
    3: [6],
    4: [],
    5: [],
    6: []
}

visited = set()

bfs(1, graph, visited)
```

```python
from collections import deque


# TC: O(V + E)
# SC: O(V)

# APPROACH:
# 1. Create Graph class using adjacency list.
# 2. add_edge() adds connection between nodes.
# 3. BFS uses queue for traversal.
# 4. Visit every node only once using visited set.


class Graph:

    def __init__(self):

        # adjacency list
        self.graph = {}

    # function to add edge
    def add_edge(self, u, v):

        # create key if not present
        if u not in self.graph:
            self.graph[u] = []

        if v not in self.graph:
            self.graph[v] = []

        # add edge
        self.graph[u].append(v)

    # BFS function
    def bfs(self, start):

        # visited set
        visited = set()

        # queue
        q = deque()

        # mark visited
        visited.add(start)

        # push start node
        q.append(start)

        while q:

            # pop front node
            node = q.popleft()

            # process node
            print(node, end=" ")

            # visit neighbours
            for adj_node in self.graph[node]:

                # visit only unvisited nodes
                if adj_node not in visited:

                    visited.add(adj_node)

                    q.append(adj_node)


# ---------------- DRIVER CODE ----------------

g = Graph()

# adding edges
g.add_edge(1, 2)
g.add_edge(1, 3)
g.add_edge(2, 4)
g.add_edge(2, 5)
g.add_edge(3, 6)

# start BFS from node 1
g.bfs(1)
```
