```python
# TC: O(V + E)
# SC: O(V)

# APPROACH:
# 1. Traverse graph using DFS.
# 2. Keep track of parent node.
# 3. If visited neighbour is not parent,
#    then cycle exists.


def dfs(node, graph, parent, visited):

    # mark visited
    visited.add(node)

    # visit neighbours
    for nei in graph[node]:

        # if neighbour not visited
        if nei not in visited:

            # current node becomes parent
            if dfs(nei, graph, node, visited):
                return True

        # visited neighbour not parent -> cycle
        elif nei != parent:
            return True

    return False

## DRIVER CODE
graph = {
    1: [2],
    2: [1, 3],
    3: [2, 4],
    4: [3, 1]
}

visited = set()

print(dfs(1, graph, -1, visited))
```

```python
# TC: O(V + E)
# SC: O(V)

# APPROACH:
# 1. Create Graph class using adjacency list.
# 2. add_edge() adds undirected edge.
# 3. DFS keeps track of parent node.
# 4. If visited neighbour is not parent,
#    then cycle exists.


class Graph:

    def __init__(self):

        # adjacency list
        self.graph = {}

    # add edge
    def add_edge(self, u, v):

        # create keys if not present
        if u not in self.graph:
            self.graph[u] = []

        if v not in self.graph:
            self.graph[v] = []

        # undirected graph
        self.graph[u].append(v)
        self.graph[v].append(u)

    # DFS cycle detection
    def dfs(self, node, parent, visited):

        # mark visited
        visited.add(node)

        # visit neighbours
        for nei in self.graph[node]:

            # if not visited
            if nei not in visited:

                # current node becomes parent
                if self.dfs(nei, node, visited):
                    return True

            # visited and not parent -> cycle
            elif nei != parent:
                return True

        return False


# ---------------- DRIVER CODE ----------------

g = Graph()

g.add_edge(1, 2)
g.add_edge(2, 3)
g.add_edge(3, 4)
g.add_edge(4, 1)

visited = set()

print(g.dfs(1, -1, visited))
```
