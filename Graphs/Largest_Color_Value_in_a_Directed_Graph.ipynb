```python
from collections import deque
from typing import List

class Solution:

    # APPROACH:
    # 1. Create graph and indegree array
    # 2. Use Kahn's Algorithm (Topological Sort using BFS)
    # 3. dp[node][c] stores:
    #       maximum frequency of color c
    #       in any path ending at node
    #
    # 4. For every edge:
    #       node -> nei
    #    propagate all color frequencies
    #
    # 5. Add current node's own color contribution
    #
    # 6. If all nodes are processed:
    #       return answer
    #    Else:
    #       cycle exists -> return -1


    # TC: O((V + E) * 26)
    #     For every edge we iterate over 26 colors
    #
    # SC: O(V * 26 + E)
    #     dp table + graph + queue


    def largestPathValue(self, colors: str, edges: List[List[int]]) -> int:

        # Number of nodes
        n = len(colors)

        # Adjacency list graph
        graph = [[] for _ in range(n)]

        # indegree[i] = incoming edges count
        indegree = [0] * n

        # Build graph
        for u, v in edges:

            # Edge: u -> v
            graph[u].append(v)

            # v has one more incoming edge
            indegree[v] += 1

        # Queue for BFS topo sort
        q = deque()

        # Push all indegree 0 nodes
        for i in range(n):

            if indegree[i] == 0:
                q.append(i)

        # dp[node][color]
        # Stores max frequency of that color
        # in any path ending at node
        dp = [[0] * 26 for _ in range(n)]

        # Final answer
        ans = 0

        # Count processed nodes
        processed = 0

        # BFS traversal
        while q:

            # Current node
            node = q.popleft()

            processed += 1

            # Convert character into index
            # a -> 0
            # b -> 1
            # ...
            color = ord(colors[node]) - ord('a')

            # Include current node's color
            dp[node][color] += 1

            # Update answer
            ans = max(ans, max(dp[node]))

            # Traverse neighbors
            for nei in graph[node]:

                # Propagate all color frequencies
                for j in range(26):

                    dp[nei][j] = max(
                        dp[nei][j],
                        dp[node][j]
                    )

                # Remove edge
                indegree[nei] -= 1

                # If indegree becomes 0
                if indegree[nei] == 0:
                    q.append(nei)

        # If all nodes processed
        # graph is acyclic
        if processed == n:
            return ans

        # Cycle exists
        return -1
```

```python
# DRIVER CODE

colors = "abaca"

edges = [
    [0,1],
    [0,2],
    [2,3],
    [3,4]
]

obj = Solution()

print(obj.largestPathValue(colors, edges))
```
