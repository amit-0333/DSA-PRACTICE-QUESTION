```python
from collections import deque


# TC: O(V + E)
# SC: O(V)

# APPROACH:
# 1. Use BFS traversal.
# 2. Store (node,parent) in queue.
# 3. If adjacent node already visited and not parent,
#    then cycle exists.


def bfs(node, graph, visited):

    # queue stores (node,parent)
    q = deque()

    # mark visited
    visited.add(node)

    # starting node parent = -1
    q.append((node, -1))

    while q:

        # pop node and parent
        node, parent = q.popleft()

        # process node
        print(node, end=" ")

        # visit neighbours
        for adj_node in graph[node]:

            # if not visited
            if adj_node not in visited:

                visited.add(adj_node)

                # current node becomes parent
                q.append((adj_node, node))

            # visited and not parent -> cycle
            elif adj_node != parent:
                return True

    return False


# ---------------- DRIVER CODE ----------------

graph = {
    1: [2, 3],
    2: [1, 4],
    3: [1, 4],
    4: [2, 3]
}

visited = set()

print("\nCycle Present:", bfs(1, graph, visited))
```
