```python
# TC: O(V + E)
# SC: O(V)

# APPROACH:
# 1. Store graph using adjacency list.
# 2. Use DFS recursion.
# 3. Visit node only once using visited set.
# 4. Print node and recursively visit neighbours.


def dfs(node, graph, visited):

    # if node not visited
    if node not in visited:

        # mark visited
        visited.add(node)

        # process current node
        print(node)

        # visit all adjacent nodes
        for adj_node in graph[node]:
            dfs(adj_node, graph, visited)


# ---------------- DRIVER CODE ----------------

# adjacency list representation
graph = {
    1: [2, 3],
    2: [4, 5],
    3: [6],
    4: [],
    5: [],
    6: []
}

# visited set
visited = set()

# start DFS from node 1
dfs(1, graph, visited)
```

```python
# TC: O(V + E)
# SC: O(V)

# APPROACH:
# 1. Create Graph class.
# 2. Store graph using adjacency list.
# 3. add_edge() function adds connection between nodes.
# 4. DFS recursively visits all unvisited neighbours.


class Graph:

    def __init__(self):

        # adjacency list
        self.graph = {}

    # function to add edge
    def add_edge(self, u, v):

        # create key if not present
        if u not in self.graph:
            self.graph[u] = []

        if v not in self.graph:
            self.graph[v] = []

        # add edge
        self.graph[u].append(v)

    # DFS function
    def dfs(self, node, visited):

        # if node not visited
        if node not in visited:

            # mark visited
            visited.add(node)

            # process node
            print(node)

            # visit neighbours
            for adj_node in self.graph[node]:
                self.dfs(adj_node, visited)


# ---------------- DRIVER CODE ----------------

g = Graph()

# adding edges
g.add_edge(1, 2)
g.add_edge(1, 3)
g.add_edge(2, 4)
g.add_edge(2, 5)
g.add_edge(3, 6)

visited = set()

# start DFS from node 1
g.dfs(1, visited)
```
