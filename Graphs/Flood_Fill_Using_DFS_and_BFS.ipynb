```python
# APPROACH:
# 1. Store original color.
# 2. Traverse connected cells using DFS.
# 3. Stop when out of bounds or color changes.
# 4. Replace with new color.

# TC: O(n * m)
# SC: O(n * m)

class Solution:

    def dfs(self, image, sr, sc, originalColor, color, n, m):

        # Directions: down, up, right, left
        direction = [(1,0), (-1,0), (0,1), (0,-1)]

        # Boundary check
        if sr < 0 or sc < 0 or sr >= n or sc >= m:
            return

        # Stop if color differs
        if image[sr][sc] != originalColor:
            return

        # Fill current cell
        image[sr][sc] = color

        # Visit neighbors
        for dr, dc in direction:
            nr = sr + dr
            nc = sc + dc

            self.dfs(image, nr, nc, originalColor, color, n, m)

    def floodFill(self, image, sr, sc, color):

        n = len(image)
        m = len(image[0])

        originalColor = image[sr][sc]

        # Avoid infinite recursion
        if originalColor == color:
            return image

        self.dfs(image, sr, sc, originalColor, color, n, m)

        return image


# ================= DRIVER CODE =================

image = [
    [1,1,1],
    [1,1,0],
    [1,0,1]
]

sr = 1
sc = 1
color = 2

obj = Solution()

result = obj.floodFill(image, sr, sc, color)

print("Updated Image:")
for row in result:
    print(row)
```

```python
from collections import deque

class Solution:

    # APPROACH:
    # 1. Store original color.
    # 2. Use queue for BFS traversal.
    # 3. Start from source cell.
    # 4. Visit all valid neighbours.
    # 5. Fill connected components with new color.

    # TC: O(n * m)
    # SC: O(n * m)

    def bfs(self, image, sr, sc, originalColor, color, n, m):

        # Directions: down, up, right, left
        direction = [(1,0), (-1,0), (0,1), (0,-1)]

        q = deque()

        # Change source color
        image[sr][sc] = color

        # Push source into queue
        q.append((sr, sc))

        while q:

            r, c = q.popleft()

            # Visit neighbours
            for dr, dc in direction:

                nr = r + dr
                nc = c + dc

                # Valid cell + same original color
                if 0 <= nr < n and 0 <= nc < m and image[nr][nc] == originalColor:

                    image[nr][nc] = color
                    q.append((nr, nc))

    def floodFill(self, image, sr, sc, color):

        n = len(image)
        m = len(image[0])

        originalColor = image[sr][sc]

        # No need to process
        if originalColor == color:
            return image

        self.bfs(image, sr, sc, originalColor, color, n, m)

        return image


# ================= DRIVER CODE =================

image = [
    [1,1,1],
    [1,1,0],
    [1,0,1]
]

sr = 1
sc = 1
color = 2

obj = Solution()

result = obj.floodFill(image, sr, sc, color)

print("Updated Image:")

for row in result:
    print(row)
```
