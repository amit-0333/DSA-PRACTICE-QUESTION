```python
from collections import deque

# TC: O(N²)
# SC: O(N²)

# APPROACH:
# 1. Use BFS because every knight move costs 1 step.
# 2. From current cell try all 8 knight moves.
# 3. Push valid unvisited cells into queue.
# 4. First time reaching target gives minimum steps.


def minSteps(n, knightPos, tarPos):

    # all 8 knight directions
    directions = [
        (-2, -1), (-2,  1),
        (-1, -2), (-1,  2),
        (1, -2),  (1,  2),
        (2, -1),  (2,  1)
    ]

    # queue -> (row, col, steps)
    q = deque()

    # visited set
    vis = set()

    row = knightPos[0]
    col = knightPos[1]

    # start BFS
    q.append((row, col, 0))

    # mark source visited
    vis.add((row, col))

    while q:

        r, c, steps = q.popleft()

        # target reached
        if [r, c] == tarPos:
            return steps

        # try all 8 moves
        for dr, dc in directions:

            nr = r + dr
            nc = c + dc

            # valid cell and unvisited
            if 0 < nr <= n and 0 < nc <= n and (nr, nc) not in vis:

                vis.add((nr, nc))

                q.append((nr, nc, steps + 1))


# ---------------- DRIVER CODE ----------------

n = 6

knightPos = [4, 5]

tarPos = [1, 1]

print(minSteps(n, knightPos, tarPos))
```

```python
from collections import deque

# TC: O(N²)
# SC: O(N²)

# APPROACH:
# 1. Use BFS because every move costs 1.
# 2. From each cell try all 8 knight moves.
# 3. Visit only valid and unvisited cells.
# 4. First time reaching target gives minimum steps.


def minSteps(n, knightPos, tarPos):

    # 8 possible knight directions
    directions = [
        (-2, -1), (-2,  1),
        (-1, -2), (-1,  2),
        (1, -2),  (1,  2),
        (2, -1),  (2,  1)
    ]

    # target row and column
    tr = tarPos[0]
    tc = tarPos[1]

    # queue -> (row, col, steps)
    q = deque()

    # visited set
    vis = set()

    sr = knightPos[0]
    sc = knightPos[1]

    # start BFS
    q.append((sr, sc, 0))

    # mark source visited
    vis.add((sr, sc))

    while q:

        r, c, steps = q.popleft()

        # target reached
        if r == tr and c == tc:
            return steps

        # explore all 8 moves
        for dr, dc in directions:

            nr = r + dr
            nc = c + dc

            # valid and unvisited cell
            if 0 < nr <= n and 0 < nc <= n and (nr, nc) not in vis:

                # mark visited
                vis.add((nr, nc))

                # push into queue
                q.append((nr, nc, steps + 1))


# ---------------- DRIVER CODE ----------------

n = 6

knightPos = [4, 5]

tarPos = [1, 1]

print(minSteps(n, knightPos, tarPos))
```

```python
from collections import deque

# TC: O(N²)
# SC: O(N²)

# APPROACH:
# 1. Use BFS level traversal.
# 2. One complete level = 1 knight move.
# 3. Process queue size at each level.
# 4. When target reached, return current steps.


def minSteps(n, knightPos, tarPos):

    directions = [
        (-2, -1), (-2,  1),
        (-1, -2), (-1,  2),
        (1, -2),  (1,  2),
        (2, -1),  (2,  1)
    ]

    sr, sc = knightPos
    tr, tc = tarPos

    q = deque()
    vis = set()

    q.append((sr, sc))
    vis.add((sr, sc))

    steps = 0

    while q:

        # current level size
        size = len(q)

        # process one level
        for _ in range(size):

            r, c = q.popleft()

            # target reached
            if r == tr and c == tc:
                return steps

            # explore all moves
            for dr, dc in directions:

                nr = r + dr
                nc = c + dc

                # valid and unvisited
                if 0 < nr <= n and 0 < nc <= n and (nr, nc) not in vis:

                    vis.add((nr, nc))

                    q.append((nr, nc))

        # one level completed
        steps += 1


# ---------------- DRIVER CODE ----------------

n = 6

knightPos = [4, 5]

tarPos = [1, 1]

print(minSteps(n, knightPos, tarPos))
```
