```python
from typing import List


# TC: O(E * (V + E))
# SC: O(V + E)

# APPROACH:
# 1. Build graph edge by edge.
# 2. After adding every edge,
#    check if cycle exists using DFS.
# 3. If cycle exists,
#    current edge is redundant.


class Solution:

    # DFS cycle detection for undirected graph
    def dfs(self, node, parent, graph, visited):

        # mark visited
        visited.add(node)

        # visit neighbours
        for nei in graph[node]:

            # if not visited
            if nei not in visited:

                if self.dfs(nei, node, graph, visited):
                    return True

            # visited and not parent -> cycle
            elif nei != parent:
                return True

        return False

    def findRedundantConnection(self, edges: List[List[int]]) -> List[int]:

        # adjacency list
        graph = {}

        # process every edge
        for u, v in edges:

            # create keys if not present
            if u not in graph:
                graph[u] = []

            if v not in graph:
                graph[v] = []

            # add edge
            graph[u].append(v)
            graph[v].append(u)

            # visited set
            visited = set()

            # if cycle found
            if self.dfs(u, -1, graph, visited):
                return [u, v]

        return []

# Example usage

sol = Solution()

edges = [
    [1, 2],
    [1, 3],
    [2, 3]
]

print(sol.findRedundantConnection(edges))
```
