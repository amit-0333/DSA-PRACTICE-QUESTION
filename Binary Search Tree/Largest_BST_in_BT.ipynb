```python
class Solution:
    def helper(self, root):
        """
        APPROACH:
        - We use postorder traversal (left → right → node).
        - For each node, we return:
            (min value in subtree, max value in subtree, size of BST in subtree)

        - Base case:
            If node is None → return (inf, -inf, 0)
            (This helps in valid comparisons)

        - For each node:
            1. Get left subtree info → (lmin, lmax, lsize)
            2. Get right subtree info → (rmin, rmax, rsize)

            3. Check BST condition:
                lmax < root.val < rmin

            4. If valid BST:
                - size = lsize + rsize + 1
                - Update global max (self.ans)
                - Return updated min and max

            5. If NOT a BST:
                - Return invalid range:
                    (-inf, inf, max(lsize, rsize))
                - This ensures parent fails BST condition

        TC:
        - O(N) → Each node is visited once

        SC:
        - O(H) → Recursion stack (H = height of tree)
        """

        if root is None:
            return (float('inf'), float('-inf'), 0)

        lmin, lmax, lsize = self.helper(root.left)
        rmin, rmax, rsize = self.helper(root.right)

        # Check BST condition
        if lmax < root.val < rmin:
            size = lsize + rsize + 1

            # Update global answer
            self.ans = max(self.ans, size)

            curr_min = min(root.val, lmin)
            curr_max = max(root.val, rmax)

            return (curr_min, curr_max, size)

        # Not a BST → return invalid range
        return (float('-inf'), float('inf'), max(lsize, rsize))

    def largestBST(self, root):
        """
        APPROACH:
        - Initialize global variable to track largest BST size
        - Call helper function
        - Return final answer

        TC:
        - O(N)

        SC:
        - O(H)
        """

        self.ans = 0
        self.helper(root)
        return self.ans
```

```python
# Definition for a binary tree node
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right


# DRIVER CODE
# Example tree:
#        5
#       / \
#      2   4
#     / \
#    1   3

root = TreeNode(5)
root.left = TreeNode(2)
root.right = TreeNode(4)
root.left.left = TreeNode(1)
root.left.right = TreeNode(3)

# Run solution
obj = Solution()
print(obj.largestBST(root))   # Expected: 3
```
