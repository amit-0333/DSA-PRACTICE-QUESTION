```python
"""
TC:
- O(N log K) → each push/pop takes log K, heap size = k+1

SC:
- O(K) → heap stores k+1 elements

APPROACH:
- Given a k-sorted array (each element is at most k distance from its correct position)
- Use a min-heap of size k+1
- Push first k+1 elements into heap
- For remaining elements:
    - Pop smallest and add to result
    - Push next element into heap
- Finally, pop all remaining elements
"""

import heapq

def sortKSorted(arr, k):

    heap = []  # min heap
    res = []   # result array

    # push first k+1 elements
    for i in range(k + 1):
        heapq.heappush(heap, arr[i])  # push into heap

    # process remaining elements
    for i in range(k + 1, len(arr)):
        res.append(heapq.heappop(heap))  # smallest element to result
        heapq.heappush(heap, arr[i])     # push next element

    # extract remaining elements
    while heap:
        res.append(heapq.heappop(heap))  # empty heap

    return res


# DRIVER CODE
# TC: O(N log K)
# SC: O(K)

arr = [7, 10, 4, 3, 20, 15]
k = 3

print(sortKSorted(arr, k))   # Expected: [3, 4, 7, 10, 15, 20]
```
