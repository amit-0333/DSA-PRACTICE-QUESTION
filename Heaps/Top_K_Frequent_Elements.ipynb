```python
from collections import Counter
import heapq
from typing import List

class Solution:
    def topKFrequent(self, nums: List[int], k: int) -> List[int]:
        """
        APPROACH:
        - Count frequency using Counter
        - Use a min-heap of size k
        - Push (frequency, number) into heap
        - If heap size exceeds k → pop smallest
        - Extract elements from heap

        TC:
        - O(N log K)

        SC:
        - O(N)
        """

        freq = Counter(nums)
        heap = []

        for num, f in freq.items():
            heapq.heappush(heap, (f, num))
            if len(heap) > k:
                heapq.heappop(heap)

        res = []
        while heap:
            res.append(heapq.heappop(heap)[1])

        return res


# DRIVER CODE
"""
APPROACH:
- Provide input array and k
- Call function
- Print result

TC:
- O(N log K)

SC:
- O(N)
"""

nums = [1, 1, 1, 2, 2, 3]
k = 2

obj = Solution()
print(obj.topKFrequent(nums, k))   # Expected: [1, 2] (order may vary)
```
