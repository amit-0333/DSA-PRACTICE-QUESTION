```python
"""
TC:
- O(N log K) → N = total elements across all lists, K = number of lists

SC:
- O(K) → heap stores one element from each list

APPROACH:
- Use a min-heap to always get the smallest current element
- Track the current maximum among elements in heap
- Initialize heap with first element of each list
- Repeatedly:
    - Pop the minimum element → defines current range with max_val
    - Update best range if smaller
    - Push next element from same list into heap
- Stop when any list is exhausted
"""

import heapq
from typing import List

class Solution:
    def smallestRange(self, nums: List[List[int]]) -> List[int]:

        heap = []  # Min-heap storing (value, list_index, element_index)
        k = len(nums)  # Number of lists

        max_val = float('-inf')  # Track current maximum value

        # Push first element of each list into heap
        for i in range(k):
            val = nums[i][0]  # First element of ith list
            heapq.heappush(heap, (val, i, 0))  # Push into heap
            max_val = max(max_val, val)  # Update max value

        st, ed = 0, float('inf')  # Store best range

        while True:
            min_val, i, j = heapq.heappop(heap)  # Get smallest element

            # Update range if better (smaller)
            if max_val - min_val < ed - st:
                st = min_val   # assign separately
                ed = max_val

            # If we reach end of any list → stop
            if j + 1 == len(nums[i]):
                break

            # Push next element from same list
            nxt_val = nums[i][j + 1]
            heapq.heappush(heap, (nxt_val, i, j + 1))

            # Update max value
            max_val = max(max_val, nxt_val)

        return [st, ed]



# DRIVER CODE
nums = [
    [4, 10, 15, 24, 26],
    [0, 9, 12, 20],
    [5, 18, 22, 30]
]

obj = Solution()
print(obj.smallestRange(nums))   # Expected: [20, 24]
```
