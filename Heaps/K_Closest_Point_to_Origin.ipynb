```python
"""
TC:
- O(N log K) → maintain heap of size k

SC:
- O(K) → heap stores k points

APPROACH:
- Use max-heap of size k (simulate using negative distance)
- Distance from origin = x^2 + y^2 (no need for sqrt)
- Push (-distance, point) into heap
- If heap size exceeds k → remove farthest point
- Extract remaining points from heap
"""

import heapq
from typing import List

def kClosest(points: List[List[int]], k: int) -> List[List[int]]:

    heap = []  # max heap using negative distance

    # iterate through points
    for i in range(len(points)):
        x, y = points[i]

        dist = x*x + y*y  # squared distance

        heapq.heappush(heap, (-dist, [x, y]))  # push point

        # maintain size k
        if len(heap) > k:
            heapq.heappop(heap)  # remove farthest point

    res = []  # result

    # extract k closest points
    while heap:
        res.append(heapq.heappop(heap)[1])

    return res


# DRIVER CODE
# TC: O(N log K)
# SC: O(K)

points = [[1, 3], [-2, 2], [5, 8], [0, 1]]
k = 2

print(kClosest(points, k))   # Example: [[-2, 2], [0, 1]] (order may vary)
```
