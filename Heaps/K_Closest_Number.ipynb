```python
"""
TC:
- O(N log K) → maintain heap of size k

SC:
- O(K) → heap stores k elements

APPROACH:
- Use a max-heap of size k (store negative distance)
- For each element:
    - Compute distance = abs(num - x)
    - Push (-distance, num) into heap
    - If heap size exceeds k → remove farthest element
- Extract elements from heap → k closest numbers
"""

import heapq

def kClosest(arr, k, x):

    heap = []  # max heap using negative distance

    # iterate through array
    for i in range(len(arr)):
        dist = abs(arr[i] - x)  # distance from x

        heapq.heappush(heap, (-dist, arr[i]))  # push (neg distance, value)

        # maintain size k
        if len(heap) > k:
            heapq.heappop(heap)  # remove farthest element

    res = []  # store result

    # extract elements
    while heap:
        res.append(heapq.heappop(heap)[1])

    return res  # k closest elements


# DRIVER CODE
# TC: O(N log K)
# SC: O(K)

arr = [5, 6, 7, 8, 9]
k = 3
x = 7

print(kClosest(arr, k, x))   # Example output: [7, 6, 8] (order may vary)
```
