```python
"""
TC:
- O(N log N) → using kthSmallest twice (each O(N log N))

SC:
- O(N) → heap usage

APPROACH:
- Find k1th smallest and k2th smallest using heap
- Iterate through array
- Sum elements strictly between these two values
"""

import heapq

def kthSmallest(arr, k):

    heap = []  # max heap using negatives

    # build heap of size k
    for i in range(len(arr)):
        heapq.heappush(heap, -arr[i])

        if len(heap) > k:
            heapq.heappop(heap)

    return -heap[0]  # kth smallest


def sumBetweenK(arr, k1, k2):

    val1 = kthSmallest(arr, k1)  # k1th smallest
    val2 = kthSmallest(arr, k2)  # k2th smallest

    total = 0

    # sum elements strictly between
    for i in range(len(arr)):
        if val1 < arr[i] < val2:
            total += arr[i]

    return total


# DRIVER CODE
# TC: O(N log N)
# SC: O(N)

arr = [1, 3, 12, 5, 15, 11]
k1 = 3
k2 = 6

print(sumBetweenK(arr, k1, k2))   # Expected: 23 (5 + 11 + 7? → depends on array)
```

```python
"""
TC:
- O(N log K) → kthLargest called twice

SC:
- O(K) → heap size

APPROACH:
- Find k1th largest and k2th largest using min-heap
- Iterate through array
- Sum elements strictly between these two values
"""

import heapq

def kthLargest(arr, k):

    heap = []  # min heap

    # build heap of size k
    for i in range(len(arr)):
        heapq.heappush(heap, arr[i])

        if len(heap) > k:
            heapq.heappop(heap)

    return heap[0]  # kth largest


def sumBetweenKLargest(arr, k1, k2):

    val1 = kthLargest(arr, k1)  # k1th largest
    val2 = kthLargest(arr, k2)  # k2th largest

    low = min(val1, val2)
    high = max(val1, val2)

    total = 0

    # sum elements strictly between
    for i in range(len(arr)):
        if low < arr[i] < high:
            total += arr[i]

    return total


# DRIVER CODE
# TC: O(N log K)
# SC: O(K)

arr = [1, 3, 12, 5, 15, 11]
k1 = 2
k2 = 4

print(sumBetweenKLargest(arr, k1, k2))   # Example output depends on ranks
```
