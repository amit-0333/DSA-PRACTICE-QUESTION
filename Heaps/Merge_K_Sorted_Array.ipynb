```python
"""
APPROACH:
- Define sample nums1 and nums2
- nums1 must have size m + n (extra space at end)
- Call merge function
- Print result

TC:
- O((m+n) log (m+n))

SC:
- O(m+n)
"""

import heapq

def merge(nums1, m, nums2, n):
    heap = []

    for i in range(m):
        heapq.heappush(heap, nums1[i])

    for i in range(n):
        heapq.heappush(heap, nums2[i])

    i = 0
    while heap:
        nums1[i] = heapq.heappop(heap)
        i += 1


# DRIVER CODE
nums1 = [1, 3, 5, 0, 0, 0]   # size = m + n
nums2 = [2, 4, 6]
m = 3
n = 3

merge(nums1, m, nums2, n)
print(nums1)   # Expected: [1, 2, 3, 4, 5, 6]
```
