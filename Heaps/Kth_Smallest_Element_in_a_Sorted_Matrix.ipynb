```python
"""
TC:
- O(k log n) → heap size = n, we do k pops

SC:
- O(n) → heap stores at most one element per row

APPROACH:
- Use min-heap to always get smallest element
- Push first element of each row into heap
- Pop k times:
    - Each pop gives next smallest element
    - Push next element from same row
- Return kth popped value
"""

import heapq
from typing import List

class Solution:
    def kthSmallest(self, matrix: List[List[int]], k: int) -> int:

        heap = []  # Min-heap storing (value, row, col)
        n = len(matrix)

        # Push first element of each row
        for r in range(n):
            heapq.heappush(heap, (matrix[r][0], r, 0))  # FIX: matrix not nums

        # Extract k smallest elements
        for _ in range(k):
            val, r, c = heapq.heappop(heap)

            # If next column exists in same row
            if c + 1 < len(matrix[0]):  # FIX: not k
                heapq.heappush(heap, (matrix[r][c + 1], r, c + 1))

        return val  # kth smallest element

## DRIVER CODE
matrix = [
    [1, 5, 9],
    [10, 11, 13],
    [12, 13, 15]
]
k = 8

obj = Solution()
print(obj.kthSmallest(matrix, k))   # Expected: 13
```
