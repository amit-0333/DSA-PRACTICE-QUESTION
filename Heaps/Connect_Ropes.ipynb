```python
"""
TC:
- O(N log N) → heap operations for N ropes

SC:
- O(N) → heap storage

APPROACH:
- Use a min-heap
- Always pick two smallest ropes
- Connect them → cost = sum
- Add cost to total
- Push combined rope back into heap
- Repeat until one rope remains
"""

import heapq

def connectRopes(arr):

    heapq.heapify(arr)  # convert to min heap

    cost = 0  # total cost

    # keep combining until one rope remains
    while len(arr) > 1:
        first = heapq.heappop(arr)   # smallest rope
        second = heapq.heappop(arr)  # second smallest rope

        curr = first + second  # cost to connect

        cost += curr  # add to total cost

        heapq.heappush(arr, curr)  # push combined rope back

    return cost


# DRIVER CODE
# TC: O(N log N)
# SC: O(N)

arr = [4, 3, 2, 6]
print(connectRopes(arr))   # Expected: 29
```
