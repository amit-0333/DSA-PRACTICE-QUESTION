```python
"""
TC:
- O(N log K) → each push/pop takes log K

SC:
- O(K) → heap stores at most k elements

APPROACH:
- Use a max heap of size k (simulate using negative values)
- Push negative of each element into heap
- If heap size exceeds k → remove largest (smallest negative)
- At the end, heap contains k smallest elements
- Top of heap gives kth smallest element
"""

import heapq

def kSmallest(arr, k):

    heap = []  # max heap using negatives

    # iterate through array
    for i in range(len(arr)):
        heapq.heappush(heap, -arr[i])  # push negative value

        # maintain heap size k
        if len(heap) > k:
            heapq.heappop(heap)  # remove largest element

    return -heap[0]  # kth smallest element


# DRIVER CODE
# TC: O(N log K)
# SC: O(K)

arr = [7, 10, 4, 3, 20, 15]
k = 3

print(kSmallest(arr, k))   # Expected: 7
```

```python
"""
TC:
- O(N log K)

SC:
- O(K)

APPROACH:
- Use a min-heap of size k
- Push elements into heap
- If heap size exceeds k → remove smallest element
- Heap will store k largest elements
- Top of heap gives kth largest element
"""

import heapq

def kLargest(arr, k):

    heap = []  # min heap to store k largest elements

    # iterate through array
    for i in range(len(arr)):
        heapq.heappush(heap, arr[i])  # push element

        # maintain heap size k
        if len(heap) > k:
            heapq.heappop(heap)  # remove smallest element

    return heap[0]  # kth largest element


# DRIVER CODE
# TC: O(N log K)
# SC: O(K)

arr = [7, 10, 4, 3, 20, 15]
k = 3

print(kLargest(arr, k))   # Expected: 10
```
