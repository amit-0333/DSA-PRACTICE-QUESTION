```python
"""
TC:
- O(N log N) → heapify O(N) + N heappop operations (log N each)

SC:
- O(N) → extra result array

APPROACH:
- Convert array into a min-heap using heapify
- Repeatedly pop smallest element
- Store popped elements in result list
- Return sorted array
"""

import heapq

def heapSort(arr):

    heapq.heapify(arr)  # Convert list into min-heap

    res = []  # Store sorted elements

    # Extract elements one by one
    while arr:
        res.append(heapq.heappop(arr))  # Get smallest element

    return res  # Return sorted array


# DRIVER CODE
arr = [1, 8, 9, 3, 4, 89, 23]
print(heapSort(arr))  # Expected: [1, 3, 4, 8, 9, 23, 89]
```

```python
"""
TC:
- O(N log N)

SC:
- O(1)

APPROACH:
- Build max heap using heapify (manual)
- Swap root (max) with last element
- Reduce heap size
- Heapify down again
"""

def heapify(arr, n, i):
    largest = i
    l = 2 * i + 1
    r = 2 * i + 2

    # check left child
    if l < n and arr[l] > arr[largest]:
        largest = l

    # check right child
    if r < n and arr[r] > arr[largest]:
        largest = r

    # swap if needed
    if largest != i:
        arr[i], arr[largest] = arr[largest], arr[i]
        heapify(arr, n, largest)


def heapSort(arr):
    n = len(arr)

    # Step 1: build max heap
    for i in range(n // 2 - 1, -1, -1):
        heapify(arr, n, i)

    # Step 2: extract elements one by one
    for i in range(n - 1, 0, -1):
        arr[0], arr[i] = arr[i], arr[0]  # move max to end
        heapify(arr, i, 0)

    return arr


# DRIVER CODE
arr = [1, 8, 9, 3, 4, 89, 23]
print(heapSort(arr))   # [1, 3, 4, 8, 9, 23, 89]
```
