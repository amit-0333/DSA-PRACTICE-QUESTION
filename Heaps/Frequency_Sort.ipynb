```python
"""
TC:
- O(N log N) → heap operations for N elements

SC:
- O(N) → Counter + heap

APPROACH:
- Count frequency using Counter
- Use max heap (via negative frequency)
- Push (−freq, element) into heap
- Pop elements and build result based on frequency
"""

from collections import Counter
import heapq

def frequencySort(arr):

    freq = Counter(arr)  # count frequency

    heap = []  # max heap using negative frequency

    # push elements into heap
    for num, f in freq.items():
        heapq.heappush(heap, (-f, num))  # (-freq, element)

    res = []  # result array

    # build result
    while heap:
        f, num = heapq.heappop(heap)  # get highest frequency

        # append element frequency times
        for _ in range(-f):
            res.append(num)

    return res


# DRIVER CODE
# TC: O(N log N)
# SC: O(N)

arr = [1, 1, 2, 2, 2, 3]
print(frequencySort(arr))   # Expected: [2, 2, 2, 1, 1, 3]
```
