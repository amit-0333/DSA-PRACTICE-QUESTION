```python
"""
TC:
- O(N log K)

SC:
- O(K)

APPROACH:
- Use a min-heap of size k
- Push each element into heap
- If heap size exceeds k → remove smallest element
- At the end, heap contains k largest elements (unsorted order)
"""

import heapq

def kLargestEle(arr, k):

    heap = []  # min heap to store k largest elements

    # iterate through array
    for i in range(len(arr)):
        heapq.heappush(heap, arr[i])  # push element into heap

        # maintain heap size k
        if len(heap) > k:
            heapq.heappop(heap)  # remove smallest element

    return heap  # returns k largest elements (not sorted)
    # return sorted(heap, reverse=True)


# DRIVER CODE
# TC: O(N log K)
# SC: O(K)

arr = [7, 10, 4, 3, 20, 15]
k = 3

print(kLargestEle(arr, k))   # Example output: [10, 15, 20] (order may vary)
```
